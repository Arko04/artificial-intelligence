#**Mathematical Definition of Nearest Neighbors**
#**The definition of nearest neighbors is based on the computation of the** **Euclidean distance from the new data point to each of the existing data**
$$ d(p,q) = \sqrt {{{({p_1} - {q_1})}^2} + {{({p_2} - {q_2})}^2} + ...{{({p_n} - {q_n})}^2}}$$

#**When we use the kNN algorithm, we need to identify the closest neighbors. Once we have identified the k neighbors that are closest to our new data point, we do not yet have a prediction. There is one step remaining to convert the multiple neighbors into one prediction. There are two prevalent methods for it.**


#**1.The first method is to take the average of the target value of the k nearest neighbors. This average is then used as the prediction**.

#**2.The second method is to take the weighted average of the k nearest neighbors and use their distances as the inverse weight so that closer points are weighted heavier in the prediction.**
#**Deciding on the Number of Neighbors k**
#**How many nearest neighbors should we include in the prediction? The value of k decides this. To apply this, let us see two cases – one nearest neighbor and three nearest neighbors – and see the difference in prediction. The two are given in the figure below**

It is typically used as a classification algorithm, working off the assumption that similar points can be found near one another.

The K-NN algorithm compares a new data entry to the values in a given data set (with different classes or categories).

Based on its closeness or similarities in a given range (K) of neighbors, the algorithm assigns the new data to a class or category in the data set (training data).

No neural networks or backpropagation

Let's break that down into steps:

1. Assign a value to K.
2. Calculate the distance between the new data entry and all other existing data entries (you'll learn how to do this shortly). Arrange them in ascending order.

3. Find the K nearest neighbors to the new entry based on the calculated distances.

4. Assign the new data entry to the majority class in the nearest neighbors.

## **Mathematical example for k = 3**
Sure! Let's go through two examples demonstrating the K-Nearest Neighbors (KNN) algorithm using both the weighted and average distance methods.

### Example 1: Weighted KNN

In weighted KNN, the contribution of each neighbor is weighted by the inverse of its distance from the query point. Closer neighbors have a larger influence on the prediction.

#### Data Points (2D space):

- Point A: (1, 2) with label 0
- Point B: (2, 3) with label 1
- Point C: (3, 1) with label 0
- Point D: (6, 5) with label 1

#### Query Point:

- Query Point Q: (3, 3)

#### Step-by-Step Calculation:

1. **Calculate distances** from Q to each data point:
- Distance to Point A: $\sqrt{(3-1)^2 + (3-2)^2} = \sqrt{4 + 1} = \sqrt{5}$
- Distance to Point B: $\sqrt{(3-2)^2 + (3-3)^2} = \sqrt{1 + 0} = 1 $
- Distance to Point C: $\sqrt{(3-3)^2 + (3-1)^2} = \sqrt{0 + 4} = 2 $
- Distance to Point D: $\sqrt{(3-6)^2 + (3-5)^2} = \sqrt{9 + 4} = \sqrt{13} $

2. **Find the 3 nearest neighbors** (k=3):
- Point B (distance = 1)
- Point C (distance = 2)
- Point A (distance = $\sqrt{5} $)

3. **Compute weights** based on distances (inverse of distance):
- Weight for Point B: $\frac{1}{1} = 1 $
- Weight for Point C: $\frac{1}{2} = 0.5 $
- Weight for Point A: $\frac{1}{\sqrt{5}} \approx 0.447 $

4. **Weighted votes**:
- Label 0: Point A (0.447) + Point C (0.5) = 0.947
- Label 1: Point B (1)

5. **Prediction**:
- Label 1 has a higher weighted vote (1 vs. 0.947), so the predicted label for Q is 1.

### Example 2: KNN with Average Distance

In the average distance method, we simply take the average of the distances of the k nearest neighbors and use it to make the prediction.

#### Data Points (2D space):

- Point A: (1, 2) with label 0
- Point B: (2, 3) with label 1
- Point C: (3, 1) with label 0
- Point D: (6, 5) with label 1

#### Query Point:

- Query Point Q: (3, 3)

#### Step-by-Step Calculation:

1. **Calculate distances** from Q to each data point:
- Distance to Point A: $\sqrt{(3-1)^2 + (3-2)^2} = \sqrt{5} $
- Distance to Point B: $\sqrt{(3-2)^2 + (3-3)^2} = 1 $
- Distance to Point C: $\sqrt{(3-3)^2 + (3-1)^2} = 2 $
- Distance to Point D: $\sqrt{(3-6)^2 + (3-5)^2} = \sqrt{13} $

2. **Find the 3 nearest neighbors** (k=3):
- Point B (distance = 1)
- Point C (distance = 2)
- Point A (distance = $\sqrt{5} $)

3. **Average distances** for each label:
- For label 0 (Points A and C):
    - Average distance = $\frac{\sqrt{5} + 2}{2} \approx \frac{2.236 + 2}{2} = 2.118 $
- For label 1 (Point B):
    - Average distance = 1

4. **Prediction**:
- The label with the smallest average distance is 1 (average distance = 1).

In both methods, the prediction for the query point Q is label 1, but they use different approaches to reach this conclusion.

#**Support Vector Machines Algorithm**

#**1.The fundamental idea behind Support Vector Machines is to fit the largest possible street/margin between the decision boundary that separates the two classes and the training instances. When performing soft margin classification, the SVM searches for a compromise between perfectly separating the two classes and having the widest possible margin. Another key idea is to use kernels when training on nonlinear datasets.**

#**2.The decision boundary is entirely determined by the support vectors. Any instance that is not a support vector (i.e., is off the street/margin) has no influence whatsoever; you could remove them, add more instances, or move them around, and as long as they stay off the street, they will not affect the decision boundary. Computing the predictions only involve the support vectors, not the whole training set.**

#**3.SVMs will fit the largest possible “street” between the classes (see the first answer), so if the training set is not scaled, the SVM will tend to neglect small features (see Figure below this cell).**

#**4.The SVM classifier can output the distance between the test instance and the decision boundary. We can use this as a confidence score. However, this score cannot be directly converted into an estimation of the class probability. If you set probability=True when creating an SVM in Scikit-Learn, then after training, it will calibrate the probabilities using Logistic Regression on the SVM’s scores (trained by an additional five-fold cross-validation on the training data). This will add the predict_proba() and predict_log_proba() methods to the SVM.**



Unscaled data can lead to models giving more weight to features with larger scales, even if they aren't necessarily more important. This can bias the model towards those features and hinder its ability to learn from other features with smaller scales.
Scaling puts all features on a similar range, allowing the model to focus on the underlying relationships between the features and the target variable, potentially leading to better performance and accuracy.

#**Soft Margin Classification**
#**If we strictly impose that all instances must be off the street and on the right side, this is called hard margin classification. There are two main issues with hard margin classification. First, it only works if the data is linearly separable. Second, it is sensitive to outliers. The above figure(SVM-1) shows the iris dataset with just one additional outlier: on the left, it is impossible to find a hard margin; on the right, the decision boundary ends up very different from the one in the Figure below (SVM-2) without the outlier, and it will probably not generalize as well. To avoid these issues, we use a more flexible model. The objective is to find a good balance between keeping the street as large as possible and limiting the margin violations (i.e., instances that end up in the middle of the street or even on the wrong side). This is called soft margin classification.**

# **Gradient-Boosted Tree (GBT) is another tree-based ensemble algorithm similar to Random Forest. GBTs use a technique known as boosting to create a strong learner from weak learners (shallow trees). GBTs train an ensemble of decision trees sequentially, with each succeeding tree decreasing the error of the previous tree. This is done by using the residuals of the previous model to fit the next model. This residual-correction process is performed a set number of iterations with the number of iterations determined by cross-validation until the residuals have been fully minimized.**

# **1.Import the necessary libraries for Gradient-Boosted Tree**

```python
from sklearn.ensemble import GradientBoostingRegressor
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.datasets import fetch_openml
boston = fetch_openml(name='boston', version=1)
```

```python
X = pd.DataFrame(boston.data, columns=boston.feature_names)
y = pd.Series(boston.target)
```

# **2.We do train test split**

```python
X_train, X_test, y_train, y_test = train_test_split(X, y)# test size will be set to default(0.25).
```

# **3. We use two nodes with three decision trees in GradientBoostingRegressor**

```python
regressor = GradientBoostingRegressor(
    max_depth=2,  # Set the maximum depth of each tree to 2
    n_estimators=3,  # Use 3 trees (estimators)
    learning_rate=1.0  # Set the learning rate to 1.0
)
regressor.fit(X_train, y_train) # Fit the gradient boosting model to the training data
```

# **4. We compute the errors and the best estimators.**

staged_predict method generates predictions at each stage (after each additional estimator is added).

```python
errors = [mean_squared_error(y_test, y_pred) for y_pred in regressor.staged_predict(X_test)]
best_n_estimators = np.argmin(errors)
```

```python
errors
```

```python
best_n_estimators
```

# **5. We use the best estimators and create our model.**




```python
best_regressor = GradientBoostingRegressor(
    max_depth=2,  # Set the maximum depth of each tree to 2
    n_estimators=best_n_estimators + 1,  # Use the optimal number of estimators found
    learning_rate=1.0  # Set the learning rate to 1.0
)
best_regressor.fit(X_train, y_train)   # Fit the new gradient boosting model to the training data
```

# **6. We predict our test set**


```python
y_pred = best_regressor.predict(X_test)
```

# **7. compute the mean absolute error.**

```python
mean_absolute_error(y_test, y_pred)
```

#**Extreme Gradient Boosting(XGBoost)Algorithm**

# **XGBoost is one of the best gradient-boosted tree implementations currently available. Released on March 27, 2014, by Tianqi Chen as a research project, XGBoost has become the dominant machine learning algorithm for classification and regression. XGBoost was designed using the general principles of gradient boosting, combining weak learners into a strong learner. But while gradient-boosted trees are built sequentially – slowly learning from data to improve its prediction in succeeding iteration, XGBoost builds trees in parallel.XGBoost produces better prediction performance by controlling model complexity and reducing overfitting through its built-in regularization. XGBoost uses an approximate algorithm to find split points when finding the best split points for a continuous feature. The approximate splitting method uses discrete bins to bucket continuous features, significantly speeding up model training. XGBoost includes another tree growing method using a histogram-based algorithm which provides an even more efficient method of bucketing continuous features into discrete bins. But while the approximate method creates a new set of bins per iteration, the histogram-based approach reuses bins over multiple iterations. This approach allows for additional optimizations that are not achievable with the approximate method, such as the ability to cache bins and parent and sibling histogram subtraction. To optimize sorting operations, XGBoost stores sorted data in in-memory units of blocks. Sorting blocks can be efficiently distributed and performed by parallel CPU cores. XGBoost can effectively handle weighted data via its weighted quantile sketch algorithm, can efficiently handle sparse data, is cache-aware, and supports out-of-core computing by utilizing disk space for large datasets, so data does not have to fit in memory.**

### Summary of XGBoost:

- **XGBoost** is a leading algorithm for classification and regression.
- It enhances gradient boosting by constructing trees in parallel, boosting efficiency and speed.
- Built-in regularization controls model complexity and minimizes overfitting.
- Uses an approximate algorithm and histogram-based methods for optimal split points in continuous features, speeding up training.
- Supports weighted data, handles sparse data, is cache-aware, and uses out-of-core computing for large datasets, ensuring high efficiency and scalability.

### Steps of the XGBoost Algorithm:

1. **Initialization**: Begin with an initial prediction.

2. **Calculate Residuals**: Determine the errors between actual values and predictions.

3. **Fit a Tree**: Train a decision tree on these residuals.

   - Prune the tree.
   - Calculate the output.

4. **Update Predictions**: Adjust predictions by adding the new tree's predictions, scaled by a learning rate.

5. **Repeat**: Iteratively add trees, each trained on the residuals from previous trees' combined predictions.

6. **Regularization**: Apply regularization to manage model complexity and prevent overfitting.

7. **Stopping Criteria**: Continue until the maximum number of trees is reached or another criterion is met.

8. **Combine Trees**: Aggregate predictions from all trees for the final result.

# **4. Split our data**


```python
X = df.drop('Churn_Value', axis=1).copy()
X.head()
```

```python
y = df['Churn_Value'].copy()
y.head()
```


# **5. We format the data with One-Hot Encoding**



```python
X.dtypes
```

```python
pd.get_dummies(X, columns=['Payment_Method']).head()
```

```python
X_encoded = pd.get_dummies(X, columns=['City',
                                       'Gender',
                                       'Senior_Citizen',
                                       'Partner',
                                       'Dependents',
                                       'Phone_Service',
                                       'Multiple_Lines',
                                       'Internet_Service',
                                       'Online_Security',
                                       'Online_Backup',
                                       'Device_Protection',
                                       'Tech_Support',
                                       'Streaming_TV',
                                       'Streaming_Movies',
                                       'Contract',
                                       'Paperless_Billing',
                                       'Payment_Method'])
X_encoded.head()
```

Stratify: When you set stratify=y, the function ensures that the split maintains the same distribution of classes in both the training and test sets as in the original dataset.

```python
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, random_state=42, stratify=y)
```


# **6.We Optimize our hyperparameters by Cross-Validation and GridSearch.**


```python
param_grid = {
     'max_depth': [4], # Maximum depth of a tree. Here, it's set to try a depth of 4.
     'learning_rate': [0.1, 0.5, 0.08],  # Defining the grid of hyperparameters for learning_rate
     'gamma': [0.25], # Defining the grid of hyperparameters for gamma
     'reg_lambda': [10.0, 20, 100], # L2 regularization term on weights. Values tried are 10.0, 20, and 100.
      'scale_pos_weight': [3]}  # Balancing the positive and negative weights. Here, it's set to 3.

optimal_params = GridSearchCV(
    # The XGBoost classifier with fixed hyperparameters:
     estimator=xgb.XGBClassifier(objective='binary:logistic', # Binary classification with logistic regression
                                 eval_metric="logloss", #  Evaluation metric for log loss.
                                 seed=42, # Random seed for reproducibility
                                 subsample=0.9, #Subsample ratio of the training instances.
                                 colsample_bytree=0.5, # Subsample ratio of columns when constructing each tree.
                                 early_stopping_rounds=8, # Stops training if validation score doesn't improve after 8 rounds.
                                 use_label_encoder=False), # Disables the use of the label encoder.
     param_grid=param_grid, #The grid of hyperparameters to search.
     scoring='roc_auc',  #The scoring metric to optimize, here ROC AUC (Area Under the Receiver Operating Characteristic Curve).
     verbose=0, #Suppresses verbose output.
     n_jobs = 10,  #Number of jobs to run in parallel.
     cv = 3, #Number of cross-validation folds.
 )
```

```python
optimal_params.fit(X_train,
                    y_train,
                    eval_set=[(X_test, y_test)],
                    verbose=False)
```

```python
print(optimal_params.best_params_) # Printing the best hyperparameters found by GridSearchCV
```

#**Our best hyperparameters:**

## **gamma: 0.25**
## **learning_rate : 0.08**
## **max_depth: 4**
## **reg_lambda: 10.0**
## **scale_pos_weight: 3**



# **7. We now build our XGBoost Model**



```python
clf_xgb = xgb.XGBClassifier(seed=42,
                        objective='binary:logistic',
                        eval_metric="logloss",
                        gamma=0.25,
                        learning_rate=0.08,
                        max_depth=4,
                        reg_lambda=10,
                        scale_pos_weight=3,
                        subsample=0.9,
                        colsample_bytree=0.5,
                        use_label_encoder=False,
                         early_stopping_rounds=20,
)
clf_xgb.fit(X_train,
            y_train,
            verbose=True,
            eval_set=[(X_test, y_test)])
```

# **8. We test our XGBoost Model**

```python


ConfusionMatrixDisplay.from_estimator(clf_xgb,
                      X_test,
                      y_test,
                     values_format='d',
                      display_labels=["Stayed", "Left"])
```
